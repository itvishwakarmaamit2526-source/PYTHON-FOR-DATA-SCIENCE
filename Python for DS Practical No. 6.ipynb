6c. Filter Dataset

Use df["Column"] to access a particular column.

```python
col = df["Marks"]
print(col)
```

Use conditions such as >, <, ==, >=, and <= to filter records

```python
print("\nStudents with Marks >= 70:")
print(df[df["Marks"] >= 70])
```

Use & for AND conditions and | for OR conditions.

```python
print("\nMarks >= 70 AND Attendance >= 85:")
print(df[(df["Marks"] >= 70) & (df["Attendance"] >= 85)])
```

```python
print("\nMarks >= 80 OR Attendance >= 90:")
print(df[(df["Marks"] >= 80) | (df["Attendance"] >= 90)])
```

6d. Sort Dataset

sort_values() is used to arrange records according to a column.

```python
result = df.sort_values("Marks")
print(result)
```

ascending=True sorts data in ascending order.

```python
print("\nMarks in Ascending Order:")
print(df.sort_values("Marks", ascending=True))
```

ascending=False sorts data in descending order.


```python
print("\nMarks in Descending Order:")
print(df.sort_values("Marks", ascending=False))
```

head() can be used to display the top records after sorting.

```python
print("\nTop Records with Highest Marks:")
print(df.sort_values("Marks", ascending=False).head())
```

6e. Perform Data Analysis

max() – finds the maximum value.

```python
print("\nMaximum Marks:")
print(df["Marks"].max())
```

min() – finds the minimum value.

```python
print("\nMinimum Marks:")
print(df["Marks"].min())
```

mean() – calculates the average.

```python
print("\nAverage Marks:")
print(df["Marks"].mean())
```

sum() – calculates the total value.

```python
print("\nTotal Marks:")
print(df["Marks"].sum())
```
