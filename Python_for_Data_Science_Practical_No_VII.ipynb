a. Scatter Plot — Age and Height of 5 Students


```python
import matplotlib.pyplot as plt

age = [12, 13, 14, 15, 16]
height = [145, 150, 155, 160, 165]

plt.scatter(age, height)
plt.title("Age vs Height of Students")
plt.xlabel("Age")
plt.ylabel("Height (cm)")

plt.show()
```

b. Bar Chart — Marks of 5 Students


```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
marks = [85, 72, 90, 65, 78]

plt.bar(students, marks)
plt.title("Marks of 5 Students")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

c. Histogram — Distribution of Marks of 10 Students


```python
import matplotlib.pyplot as plt

marks = [45, 56, 67, 72, 80, 85, 90, 65, 74, 58 ]

plt.hist(marks, bins=5)
plt.title("Distribution of Students' Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

d. Pie Chart — Students in Different Courses


```python
import matplotlib.pyplot as plt

courses = ["Science", "Commerce", "Arts", "Computer"]
students = [30, 25, 20, 25]

plt.pie(students, labels=courses, autopct="%1.1f%%")
plt.title("Students in Different Courses")

plt.show()
```

e. Bar Chart — Sales of 5 Different Products

```python
import matplotlib.pyplot as plt
import numpy as np
products = ["Pen", "Book", "Bag", "Pencil", "Bottle"]
sales = [120, 80, 60, 150, 90]

plt.bar(products, sales)
plt.title("Sales of 5 Different Products")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```
